```python
# --- Setup ---
from pymongo import MongoClient
import pandas as pd
from datetime import datetime

# --- Config ---
MONGO_URI = "mongodb+srv://pandeydevendra20devops_db_user:1Devendrapandey0@deliverly.4lvw8v3.mongodb.net/"
DB_NAME = "Jawwy"
COLLECTION = "Logs"

# --- Connect ---
client = MongoClient(MONGO_URI)
collection = client[DB_NAME][COLLECTION]

# --- Date range (May 3, 2026, UTC) ---
start = datetime(2026, 5, 3, 0, 0, 0)
end   = datetime(2026, 5, 4, 0, 0, 0)

# --- Fetch data ---
cursor = collection.find({
    "created_at": {
        "$gte": start,
        "$lt": end
    }
})

data = list(cursor)

# --- Convert to DataFrame ---
df = pd.DataFrame(data)

# --- Clean Mongo _id ---
if "_id" in df.columns:
    df["_id"] = df["_id"].astype(str)

# --- Export to Excel ---
file_name = "jawwy_may_3_2026.xlsx"
df.to_excel(file_name, index=False)

print(f"✅ Exported {len(df)} records to {file_name}")
```
